```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Keeps Plotly charts interactive on the website.
pio.renderers.default = "plotly_mimetype+notebook_connected"
GOLD, SLATE, GREY = "#D99A1E", "#4F6D9A", "#8A8F98"
```

```python
# Example chart: replace with your own data
df = pd.DataFrame({"year": [2021, 2022, 2023, 2024, 2025], "value": [3.1, 6.8, 3.9, 2.4, 2.2]})
fig = px.line(df, x="year", y="value", markers=True, template="simple_white")
fig.update_traces(line_color=GOLD, line_width=3, hovertemplate="%{x}: %{y:.1f}<extra></extra>")
fig.show()
```
